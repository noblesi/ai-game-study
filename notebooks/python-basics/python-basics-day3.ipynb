```python
units = [
    {"name": "Knight", "hp": 100},
    {"name": "Archer", "hp": 80},
    {"name": "Mage", "hp": 0},
    {"name": "Healer", "hp": 50},
]

index = 0
while index < len(units):
    unit = units[index]
    print(f"{unit['name']} 체크 중 ... (HP: {unit['hp']})")
    if unit['hp'] <= 0:
        print("기절한 유닛 발견, 탐색 중단")
        break

    index += 1
print()
for unit in units:
    if unit['hp'] >= 50:
        continue

    print(f"{unit['name']} HP: {unit['hp']}")
```

```python
player = {"name": "Player", "hp": 100, "atk": 20, "def": 5}
monster = {"name": "Slime", "hp": 60, "atk": 15, "def": 2}

turn = 1

while True:
    print("=" * 30)
    print(f"[턴 {turn}]")
    print(f"{player['name']} HP: {player['hp']}, {monster['name']} HP: {monster['hp']}")

    print("1: 공격 2: 방어 3: 물약 4: 도망")
    action = input("행동을 선택하세요: ")

    potion_cnt = 3

    if action == "1":
        damage = max(player["atk"] - monster["def"], 0)
        monster["hp"] -= damage
        print(f"{player['name']}의 공격! {monster['name']}에게 {damage} 피해를 입혔습니다.")

    elif action == "2":
        print(f"{player['name']}는 몸을 낮춰 방어 태세에 들어갑니다.")
        is_defending = True

    elif action == "3":
        if potion_cnt > 0:
            potion_cnt -= 1
            player["hp"] = min(100, player["hp"] + 30)
            print(f"{player["name"]}의 체력이 30 회복되었습니다.")
        else:
            print(f"포션의 갯수가 부족합니다.")
            continue

    elif action == "4":
        print("당신은 도망쳤습니다...")
        break

    else:
        print("잘못된 입력입니다. 다시 선택하세요.")
        continue

    if monster["hp"] <= 0:
        print(f"{monster['name']}을(를) 쓰러뜨렸습니다. 전투에서 승리했습니다.")
        break

    if is_defending == True:
        damage = max(monster["atk"] - player["def"] / 2, 0)
        player["hp"] -= damage
        print(f"{monster['name']}의 공격! {player['name']}이(가) 방어 태세를 취해 {damage} 피해를 입었습니다.")
    else:
        damage = max(monster["atk"] - player["def"], 0)
        player["hp"] -= damage
        print(f"{monster['name']}의 공격! {player['name']}이(가) {damage} 피해를 입었습니다.")

    if player["hp"] <= 0:
        print(f"{player['name']}이(가) 쓰러졌습니다... 전투에서 패배했습니다.")
        break

    turn += 1
```
