```python
import copy

units = [
    {"name": "Knight", "level": 1, "hp": 100, "atk": 20},
    {"name": "Archer", "level": 1, "hp": 70, "atk": 25},
    {"name": "Mage", "level": 1, "hp": 60, "atk": 30},
]

initial_units = copy.deepcopy(units)

def print_unit_info(unit):
    if unit is None:
        print("[print_unit_info] unit이 None입니다.")
        return
    
    print(f"=== {unit['name']} ===")
    print(f"Level : {unit['level']}")
    print(f"HP    : {unit['hp']}")
    print(f"ATK   : {unit['atk']}")
    print()

def is_alive(unit):
    return unit["hp"] > 0

print("=== 살아있는 유닛 목록 ===")
for unit in units:
    if is_alive(unit):
        print_unit_info(unit)

def take_damage(unit, damage):
    print(f"[피해] {unit['name']} 이(가) {damage} 피해를 입었습니다.")        
    unit["hp"] = max(0, unit["hp"] - damage)

    if unit["hp"] == 0:
        print(f"--> {unit['name']} 은(는) 쓰러졌습니다...")
    else:
        print(f"--> 현재 HP: {unit['hp']}")
    print()

print("=== 전투 테스트 ===")
attacker = units[0]
target = units[1]

print_unit_info(attacker)
print_unit_info(target)

take_damage(target, 30)
take_damage(target, 50)

print("=== 턴 진행: 모든 유닛이 10 피해를 받습니다. ===")
for unit in units:
    take_damage(unit, 10)

print("=== 턴 종료 후 상태 ===")
for unit in units:
    print_unit_info(unit)
```
