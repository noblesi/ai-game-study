```python
import math

while True:
    print("=== 간단 계산기 ===")
    print("1) 더하기")
    print("2) 빼기")
    print("3) 곱하기")
    print("4) 나누기")
    print("5) 제곱근")
    print("6) 거듭제곱")
    print("0) 종료")

    choice = input("메뉴를 선택하세요: ")

    if choice == "0":
        print("프로그램을 종료합니다.")
        break

    if choice not in ["1", "2", "3", "4"]:
        print("잘못된 입력입니다. 다시 선택해주세요.")
        print()
        continue

    a = float(input("첫번째 숫자: "))
    b = float(input("두번째 숫자: "))

    if choice == "1":
        result = a + b
    elif choice == "2":
        result = a - b
    elif choice == "3":
        result = a * b
    elif choice == "4":
        if b == 0:
            print("0으로 나눌 수 없습니다!")
            print()
            continue
        result = a / b
    elif choice == "5":
        x = float(input("어떤 수의 제곱근이 필요합니까? (A or B)"))
        if x == a:
            result = math.sqrt(a)
        elif x == b:
            result = math.sqrt(b)
        else:
            print("잘못된 입력입니다. 다시 선택해주세요.")
            print()
            continue
    elif choice == "6":
        result =  pow(a, b)

    print("결과: ", result)
    print()
    
```

```python
units = []

def print_units(units):
    if not units:
        print("등록된 유닛이 없습니다.")
        return
    
    print("=== 유닛 목록 ===")
    for i, unit in enumerate(units):
        print(f"{i}) {unit['name']} (Lv.{unit['level']} / HP:{unit['hp']} / ATK:{unit['atk']})")
    print()

    while True:
        print("=== 유닛 관리 메뉴 ===")
        print("1) 유닛 목록 보기")
        print("2) 유닛 추가하기")
        print("3) 유닛 레벨 올리기")
        print("4) 유닛 삭제하기")
        print("0) 종료")

        cmd = input("메뉴를 선택하세요.")

        if cmd == "0":
            print("프로그램을 종료합니다.")        
            break

        elif cmd == "1":
            print_units(units)
        
        elif cmd == "2":
            name = input("유닛이름: ")
            level = int(input("레벨: "))
            hp = int(input("HP: "))
            atk = int(input("ATK: "))

            new_unit = {
                "name": name,
                "level": level,
                "hp": hp,
                "atk": atk,
            }
            units.append(new_unit)
            print(f"{name} 유닛이 추가되었습니다.\n")
        
        elif cmd == "3":
            if not units:
                print("유닛이 없습니다. 먼저 추가해주세요.\n")
                continue

            print_units(units)
            idx_text = input("레벨을 올릴 유닛 번호를 입력하세요: ")
            idx = int(idx_text)

            if idx < 0 or idx >= len(units):
                print("잘못된 번호입니다.\n")
                continue

            units[idx]["level"] += 1
            print(f"{units[idx]['name']}의 레벨이 {units[idx]['level']}로 올랐습니다!\n")
        elif cmd == "4":
            if not units:
                print("유닛이 없습니다. 먼저 추가해주세요.\n")
                continue
            
            print_units(units)
            idx_text = input("삭제할 유닛 번호를 입력하세요: ")
            idx = int(idx_text)

            if idx < 0 or idx >= len(units):
                print("잘못된 번호입니다.\n")
                continue

            units.pop(idx - 1)
            print(f"{units[idx]['name']}이 삭제되었습니다.\n")
        else:
            print("잘못된 입력입힌다. 다시 선택해주세요.\n")
    
```
