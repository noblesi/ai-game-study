```python
import copy

level = 5
name = "Knight"
hp = 100.0

stats_list = [100, 80, 60]
stats_dict = {"hp": 100, "atk": 20, "def": 10}

print(level, name, hp)
print(stats_list)
print(stats_dict)

units = [
    {"name": "Knight", "level": 1, "hp": 100, "atk": 20},
    {"name": "Archer", "level": 1, "hp": 70, "atk": 25},
    {"name": "Mage", "level": 1, "hp": 60, "atk": 30},
]

initial_units = copy.deepcopy(units)
```

```python
def find_unit(units, name):
    for unit in units:
        if unit["name"] == name:
            return unit
    return None

def print_unit_info(unit):
    if unit is None:
        print("[print_unit_info] unit이 None입니다.")
        return

    print(f"=== {unit['name']} ===")
    print(f"Level   :{unit['level']}")
    print(f"HP      : {unit['hp']}")
    print(f"ATK     : {unit['atk']}")
    print()

def print_unit_info_by_name(units, name):
    unit = find_unit(units, name)
    if unit is None:
        print(f"[print_unit_info_by_name] {name} 유닛을 찾을 수 없습니다.")
        return
    
    print_unit_info(unit)

def print_all_units(units):
    if not units:
        print("[print_all_units] 유닛 리스트가 비어있습니다.")
        return

    print("=== 모든 유닛 정보 ===")
    for i, unit in enumerate(units, start = 1):
        print(f"[{i}] {unit['name']}")
        print(f"    level   : {unit['level']}")
        print(f"    HP      : {unit['hp']}")
        print(f"    ATK     : {unit['atk']}")
        print()
```

```python
def get_average_atk(units):
    total = 0
    for unit in units:
        total += unit["atk"]
    return total / len(units)


def get_max_hp_unit(units):
    max_unit = None
    for unit in units:
        if max_unit is None or unit["hp"] > max_unit["hp"]:
            max_unit = unit
    return max_unit
```

```python

print("평균 공격력:", get_average_atk(units))

max_unit = get_max_hp_unit(units)
print("HP가 가장 높은 유닛:", max_unit["name"], max_unit["hp"])
```

```python

def level_up(unit, hp_gain = 10, atk_gain = 5):
    unit["level"] += 1
    unit["hp"] += hp_gain
    unit["atk"] += atk_gain


def level_up_all(units, hp_gain = 10, atk_gain = 5):
    for unit in units:
        unit["level"] += 1
        unit["hp"] += hp_gain
        unit["atk"] += atk_gain

def reset_unit(units, initial_units, name):
    current = find_unit(units, name)
    if current is None:
        print(f"[reset_unit] {name} 유닛을 찾을 수 없습니다.")
        return

    original = find_unit(initial_units, name)
    if original is None:
        print(f"[reset_unit] {name} 유닛의 원본 데이터를 찾을 수 없습니다.")
        return

    current.clear()
    current.update(copy.deepcopy(original))

def reset_all_units(units, initial_units):
    if len(units) != len(initial_units):
        print("[reset_all_units] 경고 : 현재 유닛 수와 초기 유닛 수가 다릅니다.")

    for i in range(min(len(units), len(initial_units))):
        units[i].clear()
        units[i].update(copy.deepcopy(initial_units[i]))

reset_unit(units, initial_units, "Knight")
for unit in units:
    print(unit["name"], "레벨:", unit["level"], "HP:", unit["hp"], "ATK:", unit["atk"])

```
