```python
def print_damage(name, damage):
    print(f"{name}이(가) {damage}의 데미지를 입었습니다.")

print_damage("슬라임", 10)
```

```python
def calc_damage(atk, bonus):
    return atk + bonus

result = calc_damage(10 ,5)
print(result)
```

```python
def print_units(units):
    if not units:
        print("등록된 유닛이 없습니다.")
        return
    
    print("=== 유닛 목록 ===")
    for i, unit in enumerate(units):
        print(f"{i}) {unit['name']} (Lv.{unit['level']} / HP:{unit['hp']} / ATK:{unit['atk']})")
    print()
```
