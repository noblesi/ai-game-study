```python
def print_units(units):
    if not units:
        print("등록된 유닛이 없습니다.")
        return
    
    print("=== 유닛 목록 ===")
    for i, unit in enumerate(units):
        print(f"{i}) {unit['name']} (Lv.{unit['level']} / HP:{unit['hp']} / ATK:{unit['atk']})")
    print()
```

```python
units = []

def manage_units(units):
    while True:
        print("=== 유닛 관리 메뉴 ===")
        print("1) 유닛 목록 보기")
        print("2) 유닛 추가하기")
        print("3) 유닛 레벨 올리기")
        print("4) 유닛 삭제하기")
        print("0) 종료")

        cmd = input("메뉴를 선택하세요: ")

        if cmd == "0":
            print("프로그램을 종료합니다.")
            break

        elif cmd == "1":
            print_units(units)

        elif cmd == "2":
            name = input("유닛 이름: ")
            level = int(input("레벨: "))
            hp = int(input("HP: "))
            atk = int(input("ATK: "))

            new_unit = {
                "name": name.
                "level": level,
                "hp": hp,
                "atk": atk,
            }
            units.append(new_unit)
            print(f"{name} 유닛이 추가되었습니다.\n")

        elif cmd == "3":
            if not units:
                print("유닛이 없습니다. 먼저 추가해주세요.\n")
                continue

            print_units(units)
            idx_text = input("레벨을 올릴 유닛 번호를 입력하세요: ")
            idx = int(idx_text)

            if idx < 0 or idx >= len(units):
                print("잘못된 번호입니다.\n")
                continue

            units[idx]["level"] += 1
            print(f"{units[idx]['name']}의 레벨이 {units[idx]['level']}로 올랐습니다.\n")

        elif cmd == "4":
            if not units:
                print("유닛이 없습니다. 먼저 추가해주세요.\n")
                continue

            print_units(units)
            idx_text = input("삭제할 유닛 번호를 입력하세요: ")
            idx = int(idx_text)

            if idx < 0 or idx >= len(units):
                print("잘못된 번호입니다.\n")
                continue

            deleted_name = units[idx]['name']
            units.pop(idx)
            print(f"{deleted_name} 유닛이 삭제되었습니다.\n")

        else:
            print("잘못된 입력입니다. 다시 선택해주세요.\n")
```

```python
def print_unit_status(unit):
    print(f"[{unit['name']}] Lv.{unit['level']} HP:{unit['hp']} ATK:{unit['atk']}")
```

```python
def player_turn(player, monster):
    while True:
        print("\n=== 플레이어 턴 ===")
        print_unit_status(player)
        print_unit_status(monster)
        print("1) 공격 2) 방어 3) 도망")
        cmd = input("행동을 선택하세요: ")

        if cmd == "1":
            damage = player["atk"]
            monster["hp"] -= damage
            print(f"{player['name']}의 공격! {monster['name']}에게 {damage} 피해를 입혔습니다.")
            if monster["hp"] <= 0:
                print(f"{monster['name']}을(를) 쓰러뜨렸습니다!")
                return False
            return True

        elif cmd == "2":
            print(f"{player['name']}이(가) 방어 태세를 취했습니다. 다음 몬스터 공격 피해가 감소합니다.")
            player["is_defending"] = True
            return True
        
        elif cmd == "3":
            print("플레이어가 도망쳤습니다!")
            return False

        else:
            print("잘못된 입력입니다. 다시 선택해주세요.")
```

```python
def monster_turn(player, monster):
    print("\n=== 몬스터 턴 ===")

    if monster["hp"] <= 0:
        return False

    damage = monster["atk"]

    if player.get("is_defending"):
        damage //= 2
        print("플레이어가 방어 중입니다! 피해가 절반으로 감소합니다.")
        player["is_defending"] = False

    player["hp"] -= damage
    print(f"{monster['name']}의 공격! {player['name']}이(가) {damage} 피해를 입었습니다. (남은 HP: {player['hp']})")

    if player["hp"] <= 0:
        print(f"{player['name']}이(가) 쓰러졌습니다...")
        return False
    
    return True
```

```python
def battle_loop(player, monster):
    print("\n=== 전투 시작 ===")

    while True:
        kepp_battle = player_turn(player, monster)
        if not keep_battle:
            break

        if monster["hp"] <= 0:
            break

        keep_battle = monster_turn(player, monster)
        if not keep_battle:
            break

    print("\n=== 전투 종료 ===")
```
